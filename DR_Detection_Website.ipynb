```python
# STEP 2: Simple working app (no model needed!)
import gradio as gr
import numpy as np
from PIL import Image

# DR severity classes
CLASSES = ['No DR', 'Mild NPDR', 'Moderate NPDR', 'Severe NPDR', 'Proliferative DR']

def predict_dr(image):
    if image is None:
        return "❌ Please upload an image first!", "No Prediction", "0%"

    # SIMULATED PREDICTION (for demo purposes)
    # In real app, this would use a trained model
    # For now, we'll generate random predictions to show the UI works

    np.random.seed(hash(str(image.size)) % 2**32)  # Consistent for same image
    prediction = np.random.rand(5)
    prediction = prediction / prediction.sum()  # Normalize to probabilities

    confidence = float(np.max(prediction)) * 100
    predicted_class = CLASSES[np.argmax(prediction)]

    # Generate recommendation
    if predicted_class == 'No DR':
        recommendation = "✅ No diabetic retinopathy detected. Continue annual screening."
    elif predicted_class == 'Mild NPDR':
        recommendation = "⚠️ Mild DR detected. Monitor blood sugar. Recheck in 6 months."
    elif predicted_class == 'Moderate NPDR':
        recommendation = "🔶 Moderate DR detected. Refer to ophthalmologist within 3 months."
    else:
        recommendation = "🚨 SEVERE DR detected. URGENT: See ophthalmologist within 1 week!"

    # Create report
    report = f"""
### 📊 Diabetic Retinopathy Report

**Predicted Severity:** {predicted_class}
**Confidence:** {confidence:.1f}%
**Recommendation:** {recommendation}

---
*Note: This is a DEMO. In production, this would use a trained AI model.*
"""

    return report, predicted_class, f"{confidence:.1f}%"

# Create Gradio interface
with gr.Blocks(title="🔬 DR Detect - Demo", theme=gr.themes.Soft()) as demo:

    gr.Markdown("""
# 🔬 DR Detect - Diabetic Retinopathy Screening (DEMO)

Upload a retinal fundus image to see how the AI screening works.

⚠️ **This is a DEMO** - predictions are simulated. In production, this would use a trained deep learning model.
""")

    with gr.Row():
        with gr.Column(scale=1):
            image_input = gr.Image(label="📷 Upload Fundus Image", type="pil", height=400)
            analyze_btn = gr.Button("🔍 Analyze Image", variant="primary", size="lg")

        with gr.Column(scale=1):
            report_output = gr.Markdown(label="📋 Report")

    with gr.Row():
        predicted_class = gr.Label(label="Predicted Severity")
        confidence = gr.Label(label="Confidence")

    # Connect button to function
    analyze_btn.click(
        fn=predict_dr,
        inputs=image_input,
        outputs=[report_output, predicted_class, confidence]
    )

    gr.Markdown("""
---
**Built for Smart India Hackathon 2026** | Problem ID: 26038

**Tech Stack:** Gradio, Python | **Demo Mode**
""")

# Launch the app
print("🚀 Launching web app...")
demo.launch(share=True)
```
